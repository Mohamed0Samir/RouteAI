1. list of strings and a number K.

```python

#Function That checks the length of each string in the list if it's greater than k and add it in an empty list and then return the empty list
def StringGreaterK(ListOfStr, K_Len):
    Emptylist=[]
    for I_Len in ListOfStr:
        if len(I_Len)>K_Len:
            Emptylist.append(I_Len)
    return Emptylist
###############################################################################################
#list1=["apple", "pie", "banana"]
#print(StringGreaterK(list1,3))
###############################################################################################
#Function the asks the user to enter a string and add it in an empty list if they entered number zero the empty list will return the entered string 
def EnterGroupOfStr():
    emptylist=[]
    while True:
        input0=input("Enter ur word if u entered 0 no word will be entered")
        if input0=='0':
            break
        emptylist.append(input0)
    return emptylist
###############################################################################################
#Main Code
list1=EnterGroupOfStr()
LenOfStrAllowed=int(input("Enter the allowed length of string: "))
K_Str=StringGreaterK(list1,LenOfStrAllowed)
print(K_Str)
```

3. Ask for an Email as an input and part it into two on order to return the username as the first part and the second part is the domain 

```python
#Slicing the email, and return a list a length of 2 the first index is the username and the second is the domain
def Email_Slicing(Email):
    AddSignIndex=Email.find("@")
    DotSignIndex=  Email.find(".", AddSignIndex)
    EmptyList=[]
    if AddSignIndex==-1 or DotSignIndex==-1:
        print("The Email was entered incorrectly")
        return None
    #UserName Entered in the email in the first index
    EmptyList.append(Email[:AddSignIndex])
    #The Domain in the second index
    EmptyList.append(Email[AddSignIndex+1:DotSignIndex])
    return EmptyList


#Ask the user to enter an email
In_Email=input("Enter the email address that u to use to register as a username: ")
if Email_Slicing(In_Email) is not None:
    print(f"The Username is {Email_Slicing(In_Email)[0]} The Domain name is {Email_Slicing(In_Email)[1]}")
```


4. sum the number on the list as an element

```python
#A function check the integer or float numbers and return the sum of these numbers
def ListOfNumber(List):
    sum=0.0
    for ele in List:
        if type(ele)== int or type(ele)== float:
            sum+=ele
    return sum
Unit= [10,"apple", 20, True, 5.5]
print("The sum of the number on this list is: ",ListOfNumber(Unit))
```

5.Restaurant Class

```python
class Restaurant():
    Items={}
    TableReserved={1:"Not Reserved",2:"Not Reserved",3:"Not Reserved",4:"Not Reserved",5:"Not Reserved",6:"Reserved"}
    def __init__(self, customerOrders=None):
        #customer order is a list
        self.cx=customerOrders if customerOrders is not None else []

    def Rest_AddItem2Menu(self,Item,price):
        self.Items.update({Item:price})
    #Other here is the list of the class Rest
    def Rest_MakeReservation(self, TableNumber):
        for TableNum,IsReserved in self.TableReserved.items():
            if TableNum==TableNumber:
                if IsReserved=="Reserved":
                    print("Sorry This table is reserved please check another one")
                    return 0
                else:
                    self.TableReserved[TableNum]="Reserved"
                    print(f"Table Number {TableNum} got reserved for u successfully!")
                    return 1
    #Order has to be a list
    def Rest_CxOrder(self, Order):
        DictLen=len(self.Items)
        for order in Order:
            count=0
            for item in self.Items:
                if order==item:
                    self.cx.append(order)
                    break
                count+=1
                if count==DictLen:
                    print(f"The order {order} isn't in our menu sorry")

    def Rest_printMenu(self):
        for item, price in self.Items.items():
            print(f"The price of the {item} is {price}")
    #Other here is the list of the class Rest
    def Rest_printTableRes(self):
        for TableNum,Status in self.TableReserved.items():
            print(f"The table Number {TableNum} is {Status}")
    def Rest_CxOrderPrint(self):
        count=1
        for order in self.cx:
            print(f"The order number {count} is {order}")
            count+=1
    def Rest_ReceiptPrint(self):
        priceo=0
        for order in self.cx:
            for item,price in self.Items.items():
                if order==item:
                    priceo+=price
        return priceo
####################################################################################################
cx1=Restaurant()
cx1.Rest_AddItem2Menu("Sausage",90)
cx1.Rest_AddItem2Menu("Grilled Chicken",100)
cx1.Rest_AddItem2Menu("Smoky Burger Pizza",110)
cx1.Rest_AddItem2Menu("Steak",150)
cx1.Rest_MakeReservation(5)
cx1.Rest_printTableRes()
cx2=Restaurant()
cx2.Rest_printMenu()
cx2.Rest_CxOrder(["Chicken", "Sausage"])
print("The total price of your order is: ", cx2.Rest_ReceiptPrint())
cx3=Restaurant()
cx3.Rest_CxOrderPrint()



    
```
