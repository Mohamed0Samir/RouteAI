1. 3*3 Array with random values between 1-100 and then normalize it

```python
np.random.seed(5)
Rad0=np.random.randint(1,101,size=(3,3))
print(Rad0)
Max= np.max(Rad0)
Min= np.min(Rad0)
Rad0=(Rad0-Min)/(Max-Min)
print("The matrix after Normalization process is \n", Rad0)
```

2. Get the closest number to the enetered Number :{)

```python
np.random.seed(4)
Rad1=np.random.randint(1,100,10)
print(Rad1)
y=int(input("Enter a number "))
yRad1=np.abs(Rad1-y)
print(yRad1,"\n",y)
print(f"The very closest number to the number {y} is ",Rad1[np.argmin(yRad1)])
```

3. The second value for each column

```python
np.random.seed(6)
Rad2=np.random.randint(1,100,size=(5,5))
print(Rad2)
Xmin=np.zeros(5, dtype="int8")
for i in range(5):
    Col_ith=Rad2[:,i]
    Col_ith[Col_ith==np.min(Col_ith)]=np.max(Col_ith)
    Xmin[i]=np.min(Col_ith)
Xmin
```

4. Sum the each Raw and check if the result if it;s even or not

```python
np.random.seed(5)
Rad4=np.random.randint(0,100,size=(6,4))
print(Rad4)
Raw_Ith=np.zeros(6,dtype="int16")
for i in range(6):
    Raw_Ith[i]=Rad4[i,:].sum()
    if (Raw_Ith[i]%2==0):
        print(f"The sum of the raw number {i} has an even number")
Raw_Ith
```

5. Change the dividers over 5 or 3 with number -1  

```python
np.random.seed(5)
Rad5=np.random.randint(1,51,30)
print(Rad5)
Rad5[(Rad5%3==0)|(Rad5%5==0)]=-1
Rad5
```

6. Get the mean for each raw and then subtract the mean from each element in the raw

```python
np.random.seed(6)
Rad6=np.random.randint(0,50,size=(3,3))
for i in range(3):
    Raw_Ith=Rad6[i,:]
    print(f"The mean for the raw number {i} is: ",np.mean(Raw_Ith))
    Rad6[i,:]-=(np.mean(Raw_Ith)).astype("int16")
Rad6
```
